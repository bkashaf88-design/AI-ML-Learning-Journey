FOR LOOP PRACTICE

```python
images = ["cat.png", "dog.png", "car.png"]

for img in images:
    print(f"Loading image: {img}")
```

```python
for i in range(0, 10, 2):
    print(f"Index: {i}")
```

```python
lables = ["cat","dog","mouse"]

for idx,lables in enumerate (lables):
    print(f"Index {idx} : Lable is {lables}")
```

```python
student = {"name": "Kashaf", "marks": 92, "status": "Pass"}

# 1. Sirf Keys par loop:
for key in student.keys():
    print(key)

# 2. Sirf Values par loop:
for val in student.values():
    print(val)

# 3. Key aur Value dono par ek saath loop:
for key, val in student.items():
    print(f"{key} -> {val}")
```

```python
data = [10,2,-1,30,5]

for val in data:
    if val==-1:
        print("Corrupted val found (-1) , skipping....")
        continue
    print(f"processing value:{val}")
```
