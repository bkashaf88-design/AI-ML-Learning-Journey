```python
correct_pin = "1234"
balance = 1000.0

for attempt in range(3):
    pin = input("Enter PIN: ")
    if pin == correct_pin:
        print("Login Successful!\n")
        break
    else:
        print(f"Wrong PIN! Attempts left: {2 - attempt}")
else:
    print("Card Blocked!")
    exit()

while True:
    print("--- ATM MENU ---")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Select option (1-4): ")

    if choice == '1':
        print(f"Current Balance: PKR {balance}\n")

    elif choice == '2':
        amount = float(input("Enter deposit amount: "))
        if amount > 0:
            balance += amount
            print(f"New Balance: PKR {balance}\n")
        else:
            print("Invalid amount!\n")

    elif choice == '3':
        amount = float(input("Enter withdraw amount: "))
        if 0 < amount <= balance:
            balance -= amount
            print(f"Remaining Balance: PKR {balance}\n")
        else:
            print("Invalid or insufficient funds!\n")

    elif choice == '4':
        print("Thank you! Goodbye.")
        break

    else:
        print("Invalid choice!\n")
```
