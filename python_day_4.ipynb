```python
count = 1
while count <= 3:
    print(count)
    count = count + 1
print("Loop Ended")
```

```python
count = 5
while count > 0:
    print(f"Countdown: {count}")
    count -= 1
print("Launch!")
```

```python
user_attempts=0
maximum_attempts=3
while user_attempts<maximum_attempts:
    print(f"{user_attempts+1}...Entering password...")
    user_attempts+=1
if user_attempts==3:
    print()
    print()
    print("No more attempts..Account locked")
```
