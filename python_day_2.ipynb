Q1:Create a variable number = -7. Write code to check:If the number is greater than 0 $\rightarrow$ print: "Positive"If the number is less than 0 $\rightarrow$ print: "Negative"Otherwise (if it is 0) $\rightarrow$ print: "Zero"

```python
number = -7
if number > 0:
    print("The number is positive")
elif number < 0:
    print("The number is negative")
else :
    print("The number is zero")
```

Create a variable num = 24. Use the modulus operator (num % 2 == 0) to check:If the number is divisible by 2 $\rightarrow$ print: "Even Number"Otherwise $\rightarrow$ print: "Odd Number"
