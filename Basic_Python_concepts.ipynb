 (Arithmetic + Comparison)

```python
a = 10
b = 3

print(a + b)
print(a // b)   #(floor division)
print(a % b)    #(remainder)
print(a ** b)   #(power)


print(a > b)
print(a == b)
```

(Logical):


```python
age = 22
has_id = True

print(age >= 18 and has_id)   #And operator
print(age < 18 or has_id)     #or operator
```
